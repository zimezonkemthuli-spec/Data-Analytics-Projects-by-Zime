### 2. Write a Python function that takes the following list of numbers and returns the third largest value. 
####    [13, 30, 2, 11, 15, 2, 30, 15]. 
### Ensure the function handles lists with duplicate values correctly e.g., if the list is [12, 12, 5, 6], the third largest is 5. [10]

```python
#Removes duplicates, handles [7,7,5,6] correctly/Getting unique values
def third_largest(numbers):
    unique_numbers = set(numbers)

    # Check if we have at least 3 unique values
    if len(unique_numbers) < 3:
        return "Not enough unique values"

    # Sort unique values in descending order
    sorted_unique = sorted(unique_numbers, reverse=True)
    return sorted_unique[2]

# Test cases from question
print(third_largest([13, 30, 2, 11, 15, 2, 30, 15])) 
```

### 5. Load the 'Insurance' dataset (sent on the group) into a dataframe called 'insurance_df'. Standardise the columns and encode the gender, smoking status columns and region using one-hot encoding. Export the resulting dataframe to an Excel file for later use. [10]

```python
#Importing libraries
import pandas as pd

# Load Insurance dataset
insurance_df = pd.read_csv('Insurance.csv')

# Standardise columns so Linear regression works, Formula: (x - mean) / std
numeric_cols = ['Age', 'BMI', 'Children', 'Charges']
for col in numeric_cols:
    mean_val = insurance_df[col].mean()
    std_val = insurance_df[col].std()
    insurance_df[col] = (insurance_df[col] - mean_val) / std_val

# Using one-hot encoding: Convert text to numbers for, gender, smoking status, region are categorical
insurance_df = pd.get_dummies(insurance_df, columns=['Sex', 'Smoking Status', 'Region'], drop_first=False)

# Export for later use
insurance_df.to_excel('insurance_cleaned.xlsx', index=False)
print(insurance_df.head())
```

### 6. Load the 'Student Performance' dataset (sent on the group) into a dataframe called 'sp_df'. The dataset contains empty cells in certain columns. Fill in the empty cells with the median of the column in which they appear for numeric cells and with the mode for non-numeric cells. Encode non-numeric columns using one-hot encoding. Export the resulting dataframe to an Excel file for later use. [10] 

```python
#Importing libraries
import pandas as pd

# Load Student Performance dataset
sp_df = pd.read_csv(r"C:\Users\ort_d\Downloads\Student Performance (6).csv")

# Filling the empty cells with median for numeric cells and with mode for non-numeric cells
for col in sp_df.columns:
    if sp_df[col].dtype == 'str':.  65 
        mode_val = sp_df[col].mode()[0]
        sp_df[col] = sp_df[col].fillna(mode_val)
    else:
        median_val = sp_df[col].median()
        sp_df[col] = sp_df[col].fillna(median_val)

# Use one-hot encoding to onvert all text columns to numbers 
sp_df = pd.get_dummies(sp_df, drop_first=True)

# Export to excel
sp_df.to_excel('student_performance_cleaned.xlsx', index=False)
print("Cleaned shape:", sp_df.shape)
print("Nulls left:", sp_df.isnull().sum().sum())
```

### 8. How average exam scores vary by where students live and by gender. [10]

```python
#Importing libraries
import pandas as pd
import matplotlib.pyplot as plt

#Load the student prformance dataset
df = pd.read_csv(r"C:\Users\ort_d\Downloads\Student Performance (6).csv")

# Calculate average exam scores by city_type
avg_by_city = df.groupby('city_type')['final_exam_score'].mean()

# Calculate average by gender
avg_by_gender = df.groupby('gender')['final_exam_score'].mean()

# Calculate average by both (city + gender)
avg_by_both = df.groupby(['city_type','gender'])['final_exam_score'].mean()

print(avg_by_city)
print(avg_by_gender)
print(avg_by_both)

# Plot 1: By city_type
plt.figure()
avg_by_city.plot(kind='bar', color='purple')
plt.title('Average Exam Score by Where Students Live')
plt.ylabel('Average Score')
plt.xlabel('City Type')
plt.xticks(rotation=0)
plt.show()

# Plot 2: By gender
plt.figure()
avg_by_gender.plot(kind='bar', color='pink')
plt.title('Average Exam Score by Gender')
plt.ylabel('Average Score')
plt.xlabel('Gender')
plt.xticks(rotation=0)
plt.show()

# Plot 3: by where students live and by gender.
avg_by_both.unstack().plot(kind='bar')
plt.title('Average Exam Score by Location and Gender')
plt.ylabel('Average Score')
plt.xlabel('City Type')
plt.xticks(rotation=0)
plt.show()
```

### 9. Build a predictive model to estimate medical insurance charges based on demographic, lifestyle, and geographic details. Training and testing sets should be split on a ratio of 80:20. Evaluate the performance of the model using an appropriate evaluation metric. [15]

```python
#Importing libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

#Load dataset
df = pd.read_csv(r"C:\Users\ort_d\Downloads\Insurance.csv")

#Encoding categories
df = pd.get_dummies(df, columns = ['Sex','Smoking Status','Region'], drop_first = True)

#Define features x and y
x = df.drop('Charges', axis=1)
y = df['Charges']

#Train on 80%, test on 20% unseen to check generalization /split 80/20 
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Build model using y = mx + c
model = LinearRegression()
model.fit(x_train, y_train)

#Start predicting
y_pred = model.predict(x_test)

# Evaluate the metrics, R2: How much variance ,MAE: Average error in same units as Charges, and RMSE
r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

print(f"R2 Score: {r2:.3f}")
print(f"MAE: {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"Coefficients (m): {model.coef_}")
print(f"Intercept (c): {model.intercept_}")
```

### 10. Visualise actual insurance charges against predicted insurance charges on a scatter plot. [10]

```python
#Import libraries
import matplotlib.pyplot as plt

# Using y_test and y_pred 
plt.figure(figsize=(7,7))
plt.scatter(y_test, y_pred, alpha=0.6, color='black')

#Use red dash lines to see how far off the model is
min_val = min(y_test.min(), y_pred.min())
max_val = max(y_test.max(), y_pred.max())
plt.plot([min_val, max_val], [min_val, max_val], 'r--', linewidth=2)

plt.title('Actual vs Predicted Insurance Charges')
plt.xlabel('Actual Charges')
plt.ylabel('Predicted Charges')
plt.show()
```
